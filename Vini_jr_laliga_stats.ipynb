```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Vinícius Júnior Laliga Data.csv",sep=';')
df
```

<h3>Let see what each columns says:
<h4>season: The  laliga season a player played in
<h4>team: Which team a laliga player played in
<h4>apps: Appearances (Note: appearances does not mean player played full 90 minutes or played as a starting player)
<h4>min: This displays the amount of minutes played [per season]
<h4>goals: Goals [per season]
<h4>assists: Assists [per season]
<h4>sp90m: Shots per 90 mins [this is important as total shots coulb be misleading, for example:<br> | Player | Minutes | Shots | Shots/90 |<br>
| A      |     900 |    30 |      3.0 |<br>
| B      |    2700 |    60 |      2.0 | ]<br>
<h4>kp90: Key passes per 90 mins (A key pass is a pass that directly leads to a teammate taking a shot, probably resulting into goal)
<h4>xG: Expected goals [per season]
<h4>xA: Expected assists [per season]
<h4>xG90: Expected goals per 90 minutes
<h4>xA90: Expected assists per 90 minutes<br>
<p>Dataset Source: https://understat.com/player/7008

<h1>Goals Per Season

Comparing seasons:

```python
plt.figure(figsize=(20,8))
sns.barplot(data=df.sort_values('season'),x='season',y='goals')
```

Comparing how he changed overtime:

```python
plt.figure(figsize=(20,8))
sns.lineplot(
    data=df.sort_values('season'),
    x='season',
    y='goals',
    markers='o'
)
```

<h1>Assists Per Season

Comparing seasons:

```python
plt.figure(figsize=(20,8))
sns.barplot(data=df.sort_values('season'),x='season',y='assists')
```

Comparing how he changed overtime:

```python
plt.figure(figsize=(20,8))
sns.lineplot(
    data=df.sort_values('season'),
    x='season',
    y='assists',
    markers='o'
)
```

<h1>xG vs Goals

```python
sns.jointplot(x='xG',y='goals',data=df,kind='reg')
```

<h1>Or we can also compare them by...

```python
plt.figure(figsize=(20,5))
plt.subplot(1,2,1)
sns.barplot(data=df.sort_values('season'),x='season',y='assists')
plt.figure(figsize=(20,5))
plt.subplot(1,2,2)
sns.barplot(data=df.sort_values('season'),x='season',y='goals')
```
