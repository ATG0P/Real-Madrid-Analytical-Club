```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Vinícius Júnior Laliga Data.csv",sep=';')
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.drop('number',axis=1,inplace=True)
df
```

```python
df.columns
```

```python
df[['season','goals','assists']]
df['g+a'] = df['goals']+df['assists']
df
```

```python
num_col = ['apps','min','goals','assists','sp90m','kp90','xG','xA','xG90','xA90','g+a']
```

<h1>Goals Per Season

```python
plt.figure(figsize=(20,8))
sns.barplot(x=df['season'],y=df['goals'])
```

<h1>Assists Per Season

```python
plt.figure(figsize=(20,8))
sns.barplot(x=df['season'],y=df['assists'])
```

<h1>xG vs Goals

```python
sns.jointplot(x='xG',y='goals',data=df,kind='reg')
```

<h1>Or we can also compare them by...

```python
plt.figure(figsize=(20,5))
plt.subplot(1,2,1)
sns.barplot(x=df['season'],y=df['xG'])
plt.figure(figsize=(20,5))
plt.subplot(1,2,2)
sns.barplot(x=df['season'],y=df['goals'])
```

<h1>xA vs Assists

```python
plt.figure(figsize=(20,8))
sns.barplot(x=df['xA'],y=df['assists'])
```

<h1>OR

```python
sns.jointplot(x='xA',y='assists',data=df)
```

<h1>Comapring both stats season wise

```python
plt.figure(figsize=(20,5))
plt.subplot(1,2,1)
sns.barplot(x=df['season'],y=df['xA'])
plt.figure(figsize=(20,5))
plt.subplot(1,2,2)
sns.barplot(x=df['season'],y=df['assists'])
```

<h1>xG90 and xA90 trends

```python
sns.jointplot(x='xG90',y='xA90',data=df,kind='reg')
```
