<h4>xG90: Expected goals per 90 minutes

<h1>xG90 and xA90 trends

```python
sns.jointplot(x='xG90',y='xA90',data=df,kind='reg')
plt.savefig('xG90_vs_xA90.png', dpi=150, bbox_inches='tight')
```

<h4>Vinícius' xG90 fluctuated considerably, reaching a peak of 0.76 in 2023/24, while his xA90 peaked at 0.34 in 2024/25. His lowest xG90 and xA90 were 0.30 and 0.16, respectively.

<h1>Key Passes per 90

```python
plt.figure(figsize=(20,8))
sns.lineplot(
    data=df.sort_values('season'),
    x='season',
    y='kp90',
    markers='o'
)
plt.savefig('kp90.png', dpi=150, bbox_inches='tight')
```

<h4>Vinícius' KP90 dropped to 1.47 in 2023/24, compared with 2.26 in 2021/22 and 2.30 in 2024/25. However, this decline in key passes coincided with his highest xG90 of 0.76, suggesting that his attacking output was more focused on finishing and direct ball progression rather than creating shots through key passes. This highlights why KP90 should be interpreted alongside other attacking metrics rather than used as a standalone measure of creativity.

<h1>Shots Per 90

```python
plt.figure(figsize=(20,8))
sns.lineplot(
    data=df.sort_values('season'),
    x='season',
    y='sp90m',
    markers='o'
)
plt.savefig('sp90.png', dpi=150, bbox_inches='tight')
```

Vinícius started his La Liga career with a **very high shots-per-90 rate of 4.41 in 2018/19**, but this dropped sharply to **2.56 in 2019/20** and reached a low of **1.83 in 2020/21**. From there, his shot volume increased again to **2.65 in 2021/22** and eventually settled at a more moderate level of around **2.4–3.7 shots per 90**, reaching **3.14 in 2025/26**. The early 4.41 figure should also be viewed in context, as he played only **858 minutes** that season.

This may reflect his transition from a young, high-volume attacking player taking many shots into a more developed role where he combined shooting with chance creation, ball progression and decision-making.

<h1>GA per Season

```python
plt.figure(figsize=(20,8))
sns.lineplot(
    data=df.sort_values('season'),
    x='season',
    y='g+a',
    markers='o'
)
plt.savefig('ga.png', dpi=150, bbox_inches='tight')
```

<h4>Vinícius' combined goal contributions increased steadily from just 2 G+A in 2018/19 to 6 in 2020/21, before a major jump to a career-high 27 G+A in 2021/22. His output then remained relatively consistent, ranging from 19 to 21 G+A between 2022/23 and 2025/26.

## Conclusion

This analysis examined Vinícius Júnior's La Liga performance across his seasons with Real Madrid, focusing on his goals, assists, expected goals (xG), expected assists (xA), and per-90 metrics.

### Key Findings

- Vinícius' goal-scoring output increased significantly compared with his early seasons.
- His highest goal contribution in a season was in 2021/22, when he recorded 17 goals and 10 assists.
- His xG and xG90 generally increased as his attacking output developed.
- Comparing goals with xG gives an indication of the seasons in which he scored above or below his expected output.
- His xA and assists provide insight into his contribution as a creator, while KP90 measures how frequently he creates shooting opportunities for teammates.
- Per-90 statistics are useful when comparing seasons because Vinícius played different numbers of minutes in different seasons.
- I also created a `g+a` feature by adding goals and assists together to represent total goal contributions.

### Final Takeaway

Overall, the data shows a clear development in Vinícius Júnior's attacking contribution over his La Liga career at Real Madrid. His progression can be observed not only through his raw goals and assists, but also through underlying metrics such as xG, xA, xG90, xA90, and key passes per 90.

### Limitations

This analysis only uses aggregated La Liga season-level data. It does not account for factors such as position, tactical role, teammates, injuries, opposition strength, or match-by-match performance.

Further analysis could use match-level or event-level data to investigate these factors in greater detail.
